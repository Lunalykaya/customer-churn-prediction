```python
mapping = {
    'ДО 10': 5,
    'ОТ 11 ДО 50': 30,
    'ОТ 51 ДО 100': 75,
    'ОТ 101 ДО 500': 300,
    'БОЛЕЕ 500': 500
}

df['employee_count_clean_num'] = df['employee_count_clean'].map(mapping)
```

```python
df['employee_count_clean_num'] = df['employee_count_clean_num'].fillna(-1)
```

```python
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.metrics import classification_report, roc_auc_score
```

# data splitting

```python
X = df.drop(columns=['target', 'employee_count_clean'])
y = df['target']
```

```python
X_train, X_test, y_train, y_test = train_test_split(
   X, y, test_size=0.2, stratify=y, random_state=42)
```

# gradient boosting

```python
gb = GradientBoostingClassifier(random_state=42)
```

```python
cv_scores = cross_val_score(gb, X_train, y_train, cv=5)
```

```python
gb.fit(X_train, y_train)
```

```python
y_pred = gb.predict(X_test)
```

```python
print(classification_report(y_test, y_pred))
```

```python
y_proba = gb.predict_proba(X_test)[:, 1]
```

```python
print("ROC AUC:", roc_auc_score(y_test, y_proba))
```

```python
for t in [0.5, 0.4, 0.3, 0.2, 0.1]:
    print(f"\n===== Threshold = {t} =====")
    y_pred_t = (y_proba >= t).astype(int)
    print(classification_report(y_test, y_pred_t))
```
